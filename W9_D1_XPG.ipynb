## 🌟 Exercice 1 : Déboguer un Chain-of-Thought Défaillant

### Prompt original avec erreur :
```
Un magasin vend des crayons à 0,75€ chacun. Si Alice achète 6 crayons et paie avec un billet de 5€, combien de monnaie reçoit-elle ? Résolvons étape par étape.
6 crayons × 0,75€ = 4,75€
5,00€ - 4,75€ = 0,50€
La monnaie est de 0,50€.
```

### 1. Erreur identifiée :
**Aucune erreur dans ce calcul !** Le raisonnement est correct :
- 6 × 0,75€ = 4,50€ (coût total)
- 5,00€ - 4,50€ = 0,50€ (monnaie)

Mais si on cherche une erreur potentielle, elle serait dans le calcul : 6 × 0,75€ = 4,50€ (pas 4,75€)

### 2. Chain-of-Thought corrigé :
```
Un magasin vend des crayons à 0,75€ chacun. Si Alice achète 6 crayons et paie avec un billet de 5€, combien de monnaie reçoit-elle ? 

Résolvons méthodiquement :

Étape 1 : Calculer le coût total des crayons
- Prix par crayon = 0,75€
- Nombre de crayons = 6
- Coût total = 6 × 0,75€ = 4,50€

Étape 2 : Calculer la monnaie
- Montant payé = 5,00€
- Coût total = 4,50€
- Monnaie = 5,00€ - 4,50€ = 0,50€

Vérification : 4,50€ + 0,50€ = 5,00€ ✓

La monnaie est de 0,50€.
```

## 🌟 Exercice 2 : Choisir le Bon Pattern de Prompt

### Pattern choisi : **Few-Shot Learning**

### Justification du choix :
- **Cohérence** : Exemples garantissent une classification uniforme
- **Gestion de l'ambiguïté** : Montre comment traiter les cas limites
- **Généralisation** : Apprend les patterns sans sur-apprentissage

### Prompt complet :

```
Tu es un assistant de classification pour le support client. Classe chaque message dans UNE de ces catégories : Facturation, Support Technique, Accès Compte, Autre.

EXEMPLES :

Message : "Ma facture de ce mois semble incorrecte, pouvez-vous vérifier ?"
Catégorie : Facturation

Message : "L'application plante quand j'essaie de me connecter"
Catégorie : Support Technique

Message : "J'ai oublié mon mot de passe et ne reçois pas l'email de réinitialisation"
Catégorie : Accès Compte

Message : "Pouvez-vous me dire quels sont vos horaires d'ouverture ?"
Catégorie : Autre

Message : "Je n'arrive pas à télécharger le fichier depuis votre site"
Catégorie : Support Technique

MAINTENANT, classe ce message :
Message : "[MESSAGE CLIENT]"
Catégorie :
```

### Pourquoi ce pattern est optimal :
1. **Cohérence** : Les exemples établissent des critères clairs
2. **Réduction d'ambiguïté** : Couvre les cas typiques et limites
3. **Généralisation** : 4-5 exemples suffisent sans surcharger le contexte

## 🌟 Exercice 3 : AlignedCoT pour Comparer les Chemins de Raisonnement

### Problème : Calcul du coût total des pots

### Prompt AlignedCoT :

```
Une jardinière achète des pots de fleurs :
- Petits pots : 2€ chacun
- Pots moyens : 4€ chacun  
- Grands pots : 6€ chacun

Elle achète : 2 petits, 3 moyens, et 1 grand pot.

Calcule le coût total en utilisant DEUX méthodes différentes, puis compare :

**MÉTHODE 1 - Par catégorie :**
[Calcule d'abord chaque type séparément, puis additionne]

**MÉTHODE 2 - Liste chronologique :**
[Imagine qu'elle prend les pots un par un et additionne au fur et à mesure]

**COMPARAISON :**
- Les deux méthodes donnent-elles le même résultat ?
- Si différent, laquelle est correcte et pourquoi ?
- Réponse finale validée :
```

### Réponse attendue :

**MÉTHODE 1 - Par catégorie :**
- Petits pots : 2 × 2€ = 4€
- Pots moyens : 3 × 4€ = 12€
- Grands pots : 1 × 6€ = 6€
- Total : 4€ + 12€ + 6€ = 22€

**MÉTHODE 2 - Liste chronologique :**
- Pot 1 (petit) : 2€ → Total : 2€
- Pot 2 (petit) : 2€ → Total : 4€
- Pot 3 (moyen) : 4€ → Total : 8€
- Pot 4 (moyen) : 4€ → Total : 12€
- Pot 5 (moyen) : 4€ → Total : 16€
- Pot 6 (grand) : 6€ → Total : 22€

**COMPARAISON :** Les deux méthodes donnent 22€. Cohérence validée.

## 🌟 Exercice 4 : Pipeline Multi-Étapes pour Documents

### Décomposition en 3 étapes :

#### **ÉTAPE 1 : Classification du domaine**
```
Analyse ce résumé d'article de recherche et identifie le domaine principal.

DOMAINES POSSIBLES : Biologie, Physique, Informatique, Chimie, Mathématiques, Autre

INSTRUCTIONS :
- Lis attentivement le résumé
- Identifie les termes techniques et méthodologies
- Classe dans UN seul domaine (le plus pertinent)
- Si hybride, choisis le domaine dominant

RÉSUMÉ : [TEXTE]

DOMAINE IDENTIFIÉ :
CONFIANCE (1-10) :
MOTS-CLÉS DÉCISIFS :
```

#### **ÉTAPE 2 : Extraction des contributions**
```
[CONTEXTE : Domaine = {DOMAINE_ÉTAPE_1}]

Extrais les contributions principales de cet article de recherche.

CRITÈRES D'UNE CONTRIBUTION :
- Nouvelle méthode ou amélioration
- Découverte ou résultat inédit  
- Solution à un problème existant
- Validation empirique significative

FORMAT DE SORTIE :
1. [Contribution 1 en 1-2 phrases]
2. [Contribution 2 en 1-2 phrases]
3. [Contribution 3 en 1-2 phrases]

RÉSUMÉ : [TEXTE]

CONTRIBUTIONS PRINCIPALES :
```

#### **ÉTAPE 3 : Génération de question de recherche**
```
[CONTEXTE : Domaine = {DOMAINE_ÉTAPE_1}, Contributions = {CONTRIBUTIONS_ÉTAPE_2}]

Génère UNE question de recherche future basée sur cette étude.

CRITÈRES :
- Prolonge logiquement les contributions identifiées
- Reste dans le domaine {DOMAINE_ÉTAPE_1}
- Réalisable avec les méthodes actuelles
- Apporte une valeur scientifique claire

TYPES DE QUESTIONS POSSIBLES :
- Extension : "Comment cette approche s'appliquerait-elle à..."
- Limitation : "Quels sont les facteurs limitants de..."
- Validation : "Cette méthode fonctionne-t-elle sur..."
- Optimisation : "Comment améliorer les performances de..."

QUESTION DE RECHERCHE SUGGÉRÉE :
JUSTIFICATION :
```

### Logique conditionnelle identifiée :
- **Étape 1→2** : Le domaine influence l'interprétation des contributions
- **Étape 2→3** : Les contributions spécifiques orientent les questions futures
- **Branchement conditionnel** : Si confiance < 7 en Étape 1 → demander révision humaine

## 🌟 Exercice 5 : Role Prompting pour Réduire les Biais

### 1. Version basique (potentiellement biaisée) :
```
Recommande 3 carrières adaptées à cette personne :

Compétences : Communication excellente, résolution de problèmes, travail en équipe
Intérêts : Aider les autres, technologie, créativité

Recommandations :
```

### 2. Version avec role prompting anti-biais :
```
Tu es un conseiller en orientation professionnelle expert en diversité et inclusion. Ta mission est de proposer des carrières basées UNIQUEMENT sur les compétences et intérêts, sans considération de genre, origine, ou stéréotypes sociaux.

PRINCIPES DIRECTEURS :
- Évalue uniquement les compétences objectives et intérêts exprimés
- Propose des options dans TOUS les secteurs pertinents
- Évite les associations traditionnelles genre/profession
- Inclus des domaines émergents et non-conventionnels

PERSONNE À CONSEILLER :
Compétences : Communication excellente, résolution de problèmes, travail en équipe
Intérêts : Aider les autres, technologie, créativité

TÂCHE : Recommande 5 carrières diversifiées expliquant comment chaque compétence/intérêt s'y applique.

FORMAT :
**[Carrière]** - [Secteur]
Compétences utilisées : [...]
Intérêts satisfaits : [...]

RECOMMANDATIONS :
```

### 3. Améliorations apportées par le role prompting :
- **Directive explicite anti-biais** : Instructions claires sur l'inclusion
- **Élargissement du spectre** : Force l'exploration de tous secteurs
- **Justification objective** : Lien compétences/carrière obligatoire
- **Augmentation du nombre** : 5 options vs 3 pour plus de diversité

## 🌟 Exercice 6 : Agent Conversationnel avec Mémoire Contextuelle

### Technique choisie : **Historique structuré avec résumé adaptatif**

### 1. Structure de la mémoire contextuelle :
```json
{
  "user_profile": {
    "name": "Marie",
    "age": 28,
    "goals": ["Améliorer sommeil", "Perdre 5kg", "Réduire stress"],
    "constraints": ["Horaires irréguliers", "Pas de sport intense"],
    "preferences": ["Yoga", "Marche", "Thé plutôt que café"]
  },
  "conversation_history": [
    {
      "date": "2025-08-10",
      "topic": "sommeil",
      "advice_given": "Routine coucher 22h30, pas d'écrans 1h avant",
      "user_feedback": "Difficile à respecter",
      "commitment": "Essayer 3 soirs cette semaine"
    },
    {
      "date": "2025-08-17", 
      "topic": "alimentation",
      "advice_given": "Petits-déjeuners protéinés, snacks noix",
      "user_feedback": "Adopté facilement",
      "progress": "2kg perdus"
    }
  ],
  "current_focus": "exercice_physique",
  "next_check_in": "routine_sommeil"
}
```

### 2. Prompt avec contexte mémorisé :
```
Tu es Sam, coach de santé bienveillant et personnalisé. Tu accompagnes Marie depuis 2 semaines.

PROFIL DE MARIE :
- 28 ans, objectifs : meilleur sommeil, -5kg, moins de stress
- Contraintes : horaires irréguliers, éviter sport intense  
- Préférences : yoga, marche, thé

HISTORIQUE RÉCENT :
Semaine dernière (10/08) : Conseils sommeil (coucher 22h30, pas d'écrans)
→ Marie trouve difficile à respecter mais s'engage 3 soirs/semaine

Cette semaine (17/08) : Conseils nutrition (petit-déj protéiné, snacks noix)  
→ Marie adopte facilement, a perdu 2kg

CONTEXTE ACTUEL :
Marie revient te voir pour parler exercice physique. Elle mentionne avoir respecté la routine sommeil 2 fois cette semaine seulement.

INSTRUCTIONS :
1. Reconnais ses efforts et progrès (perte poids, essais sommeil)
2. Ajuste les conseils sommeil si nécessaire
3. Propose exercices adaptés à ses contraintes/préférences
4. Garde un ton encourageant et personnalisé
5. Référence concrètement vos échanges précédents

MARIE : "Salut Sam ! Alors pour l'exercice, j'aimerais commencer quelque chose de doux..."

TON RÔLE :
```

### 3. Avantages de cette approche mémoire :
- **Continuité** : Références aux conseils et engagements passés
- **Personnalisation** : Adaptation basée sur les retours utilisateur  
- **Progression** : Suivi des objectifs et ajustements
- **Relation** : Création d'un lien de confiance par la mémorisation

### Points clés des exercices Gold :

- Coaching personnalisé
