🌟 Exercice 2 : analyse critique de la conception expérimentale

Pour analyser la méthodologie et l’expérimentation, les six questions clés d’un design expérimental sont appliquées.

Question de recherche.  Le papier cherche à répondre à la question : comment implémenter un service d’IA générative basé sur un modèle de langage (LLM) en utilisant des données d’entreprise, en particulier via l’architecture Retrieval Augmented Generation (RAG). Il s’agit de démontrer que la combinaison de technologies open‑source (LangChain, Chroma DB, GPT4All) et de services propriétaires (OpenAI GPT‑3.5 turbo) permet de créer un assistant conversationnel supporté par une base documentaire interne
arxiv.org
.
Type d’étude.  Il s’agit d’une étude de cas d’implémentation (« implementation case study »). Plutôt que d’un protocole expérimental contrôlé, le papier décrit pas à pas la construction du système et illustre son fonctionnement sur des exemples concrets (documents de gestion des congés et du code vestimentaire). Il n’y a pas de comparaison expérimentale rigoureuse entre plusieurs méthodes.
Variables indépendantes et dépendantes.  Les variables indépendantes sont les choix techniques : méthode d’intégration (fine‑tuning vs RAG), modèles de langage utilisés (GPT‑3.5‑turbo, GPT4All), base vectorielle (Chroma DB ou FAISS), paramètres de chunking et d’embeddings. Les variables dépendantes sont qualitatives : pertinence des réponses générées, capacité à retrouver l’information correcte et coût de mise en œuvre. Le papier mentionne que le fine‑tuning GPT‑3.5 turbo peut atteindre un coût d’environ 2,40 USD pour un fichier de 100 000 tokens
arxiv.org
 et que ses performances peuvent parfois dépasser celles de la base GPT‑4
arxiv.org
, mais aucune métrique chiffrée n’est fournie pour l’implémentation RAG elle‑même.
Jeux de données et outils.  Les données utilisées sont des documents internes fictifs d’entreprise : un manuel de gestion des congés, un document PDF sur le code vestimentaire et un fichier texte sur le calcul d’assurance
arxiv.org
. Les outils principaux incluent :
LangChain pour l’orchestration, le chargement de documents et le découpage en morceaux
arxiv.org
.
Modèles de langage : OpenAI GPT‑3.5‑turbo et GPT4All
arxiv.org
.
Base vectorielle : Chroma DB et FAISS
arxiv.org
.
Outils de programmation : Python et Google Colab
arxiv.org
.
Présence d’un contrôle ou d’une méthode de comparaison.  Le papier évoque le fine‑tuning comme alternative et mentionne que le RAG permet d’éviter les coûts élevés de fine‑tuning. Toutefois, il ne met pas en place de groupe témoin ni de baseline quantitatif. Il s’agit donc d’une étude descriptive sans comparaison systématique à un autre système ou à des métriques standard.
Transparence et reproductibilité.  La procédure est décrite étape par étape avec les bibliothèques et les étapes clés (chargement, découpage, embeddings, base vectorielle, requête et génération). Le code n’est pas fourni intégralement, mais les références aux bibliothèques (LangChain, Chroma, OpenAI API) et aux scripts d’exemple (installation, chargement de fichiers) sont détaillées
arxiv.org
. Cela permet à un lecteur de reproduire l’approche, sous réserve d’avoir accès aux mêmes outils et données. L’étude manque cependant de description des paramètres précis (par ex. tailles des embeddings, nombre de morceaux conservés) et de résultats quantitatifs, ce qui limite l’évaluation rigoureuse de la reproductibilité.


🌟 Exercice 3 : métriques d’évaluation et preuves

Le papier ne présente pas de section dédiée à l’évaluation quantitative. Les comparaisons évoquées se limitent à des affirmations générales sur les bénéfices du fine‑tuning ou du RAG :

L’introduction mentionne que le fine‑tuning de GPT‑3.5 turbo peut offrir des performances comparables, voire supérieures, au modèle de base GPT‑4 pour certaines tâches spécifiques
arxiv.org
.
Le coût approximatif d’un fine‑tuning sur 100 000 tokens (environ 2,40 USD) est indiqué
arxiv.org
, ce qui suggère un critère économique.
La limite de contexte de GPT‑3.5 et GPT‑4 (environ 8 000 mots et 50 pages) est rappelée pour motiver l’usage du RAG
arxiv.org
.
Aucune métrique de performance (précision des réponses, temps de réponse, taux de hallucination) n’est rapportée. Les preuves reposent sur des exemples qualitatifs où des questions sur le code vestimentaire sont correctement répondues grâce au RAG
arxiv.org
. Pour évaluer rigoureusement l’approche, plusieurs indicateurs pourraient être ajoutés :

Précision des réponses (proportion de réponses correctes par rapport aux documents sources).
Taux de hallucination (part des informations inventées par l’IA).
Latence de réponse (temps moyen pour générer une réponse en fonction de la taille de la base vectorielle).
Feedback humain : évaluation par des utilisateurs finaux de la pertinence et de la clarté des réponses.
Comparaison fine‑tuning vs RAG sur un même corpus : coût total, précision, flexibilité (mise à jour rapide des données), etc.

🌟 Exercice 4 : prise de notes selon la méthode Cornell (section 3.2.1)

Dans la section 3.2.1, l’auteur détaille le processus de mise en œuvre du modèle RAG. Voici des notes au format Cornell :

Colonne de gauche : mots‑clés / questions	Colonne de droite : notes synthétiques
Collection de données	Réunir données structurées (CSV, JSON, XML) et non structurées (PDF, TXT, images, vidéos). Utiliser LangChain pour charger manuels, règlements ou pages web
arxiv.org
.
Découpage en fragments (chunking)	Diviser les documents en morceaux (phrases ou paragraphes). LangChain fournit des fonctions pour segmenter en fragments adaptés à la recherche
arxiv.org
.
Embeddings	Transformer chaque fragment en vecteur numérique via des bibliothèques OpenAI ou GPT4All
arxiv.org
.
Base de données vectorielle	Construire une base (Chroma ou FAISS) stockant les embeddings et les documents, permettant la recherche par similarité
arxiv.org
.
Intégration de la requête et des résultats de recherche	Convertir la question en embedding, rechercher les morceaux les plus similaires dans la base, puis transmettre ces morceaux au LLM via LangChain
arxiv.org
.
Génération de la réponse	Le LLM (GPT‑3.5 turbo ou GPT4All) utilise les fragments récupérés comme contexte pour produire une réponse pertinente
arxiv.org
.
Résumé : La procédure RAG intégrée à LangChain commence par recueillir des données variées et les charger dans un pipeline. Chaque document est découpé en fragments puis transformé en vecteurs via des embeddings. Ces vecteurs sont stockés dans une base vectorielle (Chroma ou FAISS). Lorsque l’utilisateur pose une question, la requête est également vectorisée pour chercher les fragments les plus proches. Ces fragments sont fournis au LLM qui génère une réponse contextualisée, limitant les hallucinations et s’appuyant sur des informations à jour
arxiv.org
.

🌟 Exercice 6 : réflexion sur la conception et application personnelle

Pour appliquer les enseignements du papier, imaginons un cas d’usage : assistant de connaissance RH destiné aux employés d’une entreprise multinationale. Cet assistant répond aux questions sur les politiques internes (congé, avantages, code éthique) dans différentes langues.

Collecte et préparation des données.
Rassembler les politiques RH, manuels du personnel, FAQ, accords collectifs et mises à jour légales au format PDF, DOCX et HTML.
Nettoyer ces documents (suppression des doublons, anonymisation) et les diviser en fragments cohérents (par exemple, 300‑500 mots).
Veiller à actualiser régulièrement le corpus (nouveaux accords ou lois).
Choix de la base vectorielle.
Chroma DB convient pour un prototype grâce à sa simplicité d’utilisation et à son stockage en mémoire, comme dans l’étude
arxiv.org
.
Pour une solution en production avec de gros volumes, FAISS ou Pinecone pourraient être préférés pour leur scalabilité et leur persistance. La décision dépendra du budget et des contraintes de confidentialité.
Intégration du LLM.
Utiliser un modèle commercial performant, par exemple GPT‑4 (capacité contextuelle de 128 000 tokens) via OpenAI ou Llama 2 déployé en interne pour des raisons de confidentialité.
Compléter par des modèles open‑source (GPT4All, Mistral) pour des tâches spécifiques ou hors ligne.
Personnaliser les messages système pour rappeler au modèle de citer les extraits de documents et de rester neutre.
Gestion des hallucinations et des réponses obsolètes.
Appliquer le RAG : la question de l’utilisateur est convertie en vecteur, les fragments pertinents sont extraits puis injectés dans le prompt pour que le LLM se base sur des sources vérifiables.
Implémenter un filtre de confiance : calculer la similarité moyenne des fragments récupérés; si elle est trop faible, inviter l’utilisateur à reformuler ou indiquer que l’information n’est pas disponible.
Mettre à jour régulièrement la base documentaire et inclure des horodatages; ajouter un module qui avertit lorsque des politiques ont changé.
Défis anticipés.
Sécurité et confidentialité : les documents RH contiennent des informations sensibles; il faut chiffrer la base vectorielle et restreindre l’accès.
Maintenance des données : maintenir à jour les documents et gérer la suppression ou l’archivage de politiques obsolètes.
Qualité des réponses : calibrer le découpage pour ne pas perdre de contexte; ajuster les paramètres du modèle (température, max_tokens).
Multilinguisme : indexer des documents en plusieurs langues et utiliser des embeddings multilingues; gérer les différences juridiques selon les pays.
Coût : les modèles propriétaires peuvent être onéreux; recourir à un mélange de modèles open‑source et payants permet d’optimiser le budget.
En suivant cette approche, un assistant RH fondé sur une architecture RAG + LLM peut fournir des réponses fiables et actualisées tout en respectant la confidentialité des données. L’intégration du RAG permet de réduire les hallucinations en s’appuyant sur les documents internes, et la modularité du système facilite son adaptation à d’autres domaines (assistance juridique, support technique, formation).
