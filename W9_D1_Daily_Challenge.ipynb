# Exercice : Techniques de Prompting Avancées

## 👩‍🏫 Ce que vous apprendrez
- Comment appliquer des techniques de prompting avancées pour façonner le ton, la structure et le format du contenu généré par l'IA
- Comment contraindre les réponses de l'IA en utilisant des limites de mots, de phrases et des formats de sortie
- Comment guider le modèle pour paraphraser, citer ou résumer un texte avec précision
- Comment évaluer et réviser les prompts pour de meilleurs résultats
- Comment détecter et atténuer le contenu halluciné ou fabriqué

## 🛠️ Ce que vous créerez
Vous concevrez, testerez et affinerez un prompt qui effectue **quatre tâches de contrôle différentes** dans un flux de travail réaliste de génération de contenu. Vous travaillerez à travers un cycle complet de développement de prompt :
1. Création d'un prompt avec **contrôle du ton, de la structure, du format et de la longueur**
2. Évaluation de la sortie pour la qualité et la précision
3. Atténuation des risques d'hallucination
4. Amélioration de la clarté grâce aux techniques de citation ou de paraphrase

## Ce que vous utiliserez
- **Concepts :** prompting avancé (ton, structure, format, longueur), détection d'hallucination, paraphrase vs citation
- **Techniques :** prompting par rôle, prompting basé sur les instructions, contraintes de sortie, évaluation et révision itératives
- **Méthode :** cycle de développement de prompt (concevoir → générer → évaluer → affiner)
- **Modèle :** IA générative (ex. ChatGPT/GPT-4, Gemini, DeepSeek, …)
- **Dataset :** le paragraphe de politique fourni sur la sécurité d'accès distant

## Scénario
Vous travaillez dans l'**Équipe Apprentissage & Communications** d'une entreprise technologique mondiale. On vous a demandé d'utiliser l'IA générative pour créer du **contenu de micro-apprentissage destiné aux employés** pour un bulletin interne.

Votre tâche est de transformer un paragraphe d'une **récente mise à jour de politique interne** en un **extrait de connaissances** concis et clair. Le texte de politique original est trop formel, dense et plein de jargon.

# Instructions et Solutions

## Étape 1 : Créer un prompt qui contrôle le ton, le format et la longueur de la sortie

### Texte d'entrée original :
*Les employés doivent s'assurer que tout accès distant aux systèmes internes soit établi via le VPN sécurisé approuvé. En aucune circonstance, des connexions non sécurisées ou des appareils personnels dépourvus de protection des points d'accès ne doivent être utilisés pour accéder aux données propriétaires ou aux communications sensibles.*

### Prompt créé :
```
Réécris le texte de politique suivant pour en faire une communication à destination de l'employé conviviale et claire. Formate ta réponse sous forme de puces, paraphrase le contenu (ne cite pas directement), et garde la réponse totale sous 75 mots. Rends-le accessible à tous les employés, quel que soit leur niveau technique.

Texte d'entrée : [texte original]

Exigences :
- Ton convivial et clair
- Format en puces
- Contenu paraphrasé uniquement
- Moins de 75 mots au total
- Facile à comprendre pour le personnel non technique
```

### Sortie générée :
• Connectez-vous toujours aux systèmes de l'entreprise via notre VPN approuvé quand vous travaillez à distance
• N'utilisez jamais de connexions internet non sécurisées pour accéder aux données de travail
• Assurez-vous que vos appareils personnels ont un logiciel de sécurité approprié avant d'accéder aux informations de l'entreprise
• Aidez à protéger les communications sensibles et les données de l'entreprise contre les menaces de sécurité

**Nombre de mots :** 64 mots

---

## Étape 2 : Évaluer la sortie

| **Critères**   |   **Questions d'orientation**         |                    **Évaluation**       |
|--------------  |---------------------------------------|-----------------------------------------
| **Pertinence** | La sortie respecte-t-elle le message du contenu source ? | ✅ **Oui** - 
Tous les points clés sont conservés : VPN obligatoire, interdiction des connexions non sécurisées, protection des appareils |
| **Clarté** | Le message est-il facile à comprendre pour le personnel non technique ? | ✅**Oui** - Langage simplifié, termes techniques expliqués |
| **Structure** | L'information est-elle formatée en puces claires ? |               ✅ **Oui** 
- 4 puces bien organisées, progression logique |
| **Ton** | Le langage est-il approprié pour la communication interne aux employés ? | ✅ **Oui** - Ton convivial, utilisation du "vous", langage accessible |
| **Longueur** | Reste-t-elle sous 75 mots ?                                          | ✅ **Oui** 
- 58 mots, bien en dessous de la limite |
| **Précision factuelle** | Le modèle a-t-il inventé ou supprimé des détails de sécurité critiques ? | 
⚠️ **Attention** - Ajout potentiel : "logiciel de sécurité approprié" pourrait être une hallucination |

**Résultat global :** La sortie est globalement réussie, mais il y a un risque mineur d'hallucination.

## Étape 3 : Détecter et atténuer les hallucinations

### Hallucination détectée :
Le modèle a ajouté "logiciel de sécurité approprié" alors que le texte original mentionne seulement "protection des points d'accès".

### Prompt révisé :
```
Réécris le texte de politique suivant pour en faire une communication employé conviviale et claire. Formate ta réponse sous forme de puces, paraphrase le contenu (ne cite pas directement), et garde la réponse totale sous 75 mots.

IMPORTANT : Paraphrase UNIQUEMENT le contenu du texte d'entrée. N'ajoute aucune nouvelle recommandation, technologie ou détail qui ne figure pas explicitement dans le texte original.

Texte d'entrée : [texte original]

Exigences :
- Ton convivial et clair
- Format en puces
- Paraphrase uniquement ce qui est dans le texte original
- Moins de 75 mots au total
```

### Nouvelle sortie (plus fidèle) :
• Utilisez toujours le VPN sécurisé approuvé pour accéder aux systèmes de l'entreprise à distance
• N'utilisez jamais de connexions non sécurisées pour ce type d'accès
• Les appareils personnels sans protection appropriée ne doivent pas servir à consulter nos données
• Cette règle s'applique aux données propriétaires et communications sensibles

**Nombre de mots :** 49 mots

## Étape 4 : Paraphrase pour un public de stagiaires juniors

### Prompt pour stagiaires :
```
Réécris ce texte de politique de sécurité pour des stagiaires juniors qui débutent dans l'entreprise. Utilise un langage simple et des phrases courtes. Formate en maximum 4 puces. Évite tout jargon corporatif ou légal. Garde un ton de soutien et informatif, comme si tu expliquais à quelqu'un qui découvre ces règles pour la première fois.

Texte d'entrée : [texte original]

Exigences :
- Maximum 4 puces
- Langage très simple
- Phrases courtes
- Aucun jargon technique ou corporatif
- Ton de soutien
```

### Sortie pour les stagiaires :
• **Travail à distance :** Utilise toujours le VPN de l'entreprise pour te connecter
• **Wifi public :** Évite les connexions non protégées (café, aéroport, etc.)
• **Ton téléphone/ordinateur perso :** Assure-toi qu'il est bien protégé avant de l'utiliser pour le travail
• **Pourquoi c'est important :** Cela protège les informations confidentielles de l'entreprise

**Caractéristiques :** Langage familier, exemples concrets, explication du "pourquoi", ton rassurant et éducatif.

## Étape 5 : Variante d'extraction de citation

### Prompt pour extraire une citation :
```
Lis le texte de politique suivant et extrais UNE citation directe qui capture le mieux la politique de sécurité principale. Choisis la phrase ou le segment qui résume l'essentiel de la règle.

Texte d'entrée : [texte original]

Fournis la citation exacte entre guillemets.
```

### Citation extraite :
*"Les employés doivent s'assurer que tout accès distant aux systèmes internes soit établi via le VPN sécurisé approuvé."*

### Analyse : Quand utiliser la citation vs la paraphrase ?

#### 🔹 Quand la citation est plus appropriée :
- **Communications légales/conformité** : Bulletins de politique officielle, formations de conformité
- **Mémos exécutifs** : Quand la précision juridique est cruciale
- **Documentation d'audit** : Pour traçabilité et responsabilité
- **Communications de crise** : Pour éviter les malentendus sur des règles critiques

#### 🔹 Quand la citation pose des risques :
- **Barrière linguistique** : Le jargon peut bloquer la compréhension
- **Résistance des employés** : Ton trop autoritaire ou légaliste
- **Contexte inapproprié** : Dans des supports informels ou éducatifs
- **Surcharge d'information** : Quand la citation est trop dense ou technique

#### 💡 Recommandation :
Utilisez la citation pour l'autorité et la précision, la paraphrase pour l'accessibilité et l'engagement.

## Conclusion

Cet exercice démontre l'importance du contrôle précis des prompts pour :
- Adapter le contenu à différents publics
- Maintenir la fidélité au contenu source
- Détecter et corriger les hallucinations
- Choisir la bonne stratégie (paraphrase vs citation) selon le contexte

Les techniques apprises ici sont essentielles pour un usage professionnel efficace de l'IA générative dans la communication d'entreprise.
