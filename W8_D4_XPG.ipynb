3 Critique de la conception expérimentale (200 mots maximum)
La section 6 évalue NExT‑GPT sur une large gamme de tâches de perception et de génération afin de justifier son caractère any‑to‑any. Les auteurs utilisent des jeux de données publics : NoCaps, Flickr30K et COCO pour la légende d’images ; VQAv2, VizWiz et OKVQA pour le question‑réponse visuel ; MMB et SEED pour des benchmarks de compréhension pure ; MSR‑VTT, MSVD‑QA, MSRVTT‑QA et NExTQA pour la vidéo ; AudioCaps pour l’audio. Des tableaux comparatifs montrent les performances du modèle par rapport à des baselines reconnues (InstructBLIP, LLaVA, Emu, Codi, UIO‑2XXL, etc.)
arxiv.org
arxiv.org
. Les métriques utilisées sont appropriées : CIDEr pour la caption, précision pour le QA, FID/FAD/CLIPSIM pour la génération et une évaluation humaine pour l’aspect qualitatif
arxiv.org
arxiv.org
. Toutefois, certaines limitations apparaissent : (i) les performances zero‑shot sont rapportées mais la procédure de fine‑tuning sur les datasets d’alignement n’est pas entièrement détaillée (risque d’overfitting) ; (ii) le jeu MosIT, essentiel pour le raisonnement, est interne et non ouvert, ce qui empêche la reproduction ; (iii) certaines baselines (p.ex. InstructBLIP ou Emu) ne sont pas nécessairement optimisées pour toutes les modalités, tandis que d’autres modèles plus récents ne sont pas comparés ; (iv) les biais des jeux de données (images occidentales, instructions générées par GPT‑4) peuvent se refléter dans les résultats. Malgré ces réserves, les expériences démontrent un bon support pour les revendications principales et illustrent clairement la flexibilité any‑to‑any du système.

4 Glossaire des métriques quantitatives
 Métrique	 Ce qu’elle mesure	 Usage dans les tâches multimodales	 Limites
CIDEr (Consensus‑based Image Description Evaluation)	Calculé à partir de la similarité TF–IDF entre la légende générée et les légendes de référence.	Évalue la qualité des légendes d’images ou de vidéos en se basant sur la concordance avec plusieurs références. Utilisé pour les tâches de captioning dans les tables 2 et 3
arxiv.org
.	Sensible au choix des références ; favorise des mots rares sur l’alignement sémantique ; peut être trompeur pour des descriptions valides mais différentes.
Précision / Exactitude	Taux de réponses correctes sur des jeux de questions visuelles ou vidéo.	Mesure la capacité de question‑réponse (VQAv2, VizWiz, OKVQA, MSVD‑QA, MSRVTT‑QA, NExTQA).	Ne prend pas en compte la qualité de l’explication ni le raisonnement ; sensible aux biais de questions.
FID (Fréchet Inception Distance)	Distance entre les distributions des embeddings d’images générées et de vraies images (via un réseau Inception).	Utilisé pour évaluer la qualité et la diversité de la génération d’images et de vidéos (plus le score est faible, meilleure est la qualité). NExT‑GPT obtient un FID = 10,07 pour les images et 11,18 en zéro‑shot
arxiv.org
.	Suppose que les embeddings d’Inception sont adaptés pour toutes les modalités visuelles ; insensible à la cohérence sémantique avec le texte.
FAD (Fréchet Audio Distance)	Distance entre les distributions des embeddings audio calculées avec un modèle VGGish.	Évalue la qualité des sons générés ; plus la valeur est basse, plus le son est réaliste. Le modèle atteint 1,68
arxiv.org
.	Basé sur un espace d’embedding particulier ; ne reflète pas toujours la perception humaine (qualité, bruit).
CLIPSIM (CLIP Similarity)	Similarité cosinus entre l’embedding CLIP de la description textuelle et celui du média généré.	Utilisé pour vérifier si les images, vidéos ou sons générés correspondent au contenu textuel (score plus élevé indiquant une meilleure correspondance)
arxiv.org
.	Dépend du modèle CLIP et favorise la similarité visuelle ; ne mesure pas la qualité perceptive ou artistique.
Évaluation humaine	Notes subjectives sur le suivi d’instructions, la rationalité et la qualité (sur 1–100).	Permet d’évaluer la pertinence globale du système sur des tâches complexes (100 instructions). NExT‑GPT obtient des scores supérieurs à Visual‑ChatGPT et HuggingGPT
arxiv.org
.	Échantillon limité ; biais potentiels des évaluateurs ; absence de protocole détaillé.
5 Bullet Bank : 10 insights clés
Système any‑to‑any complet : NExT‑GPT connecte un LLM (Vicuna‑7B) à des encodeurs multimodaux (ImageBind) et à des décodeurs de diffusion (Stable Diffusion, Zeroscope, AudioLDM). Cette architecture permet de traiter et de produire du texte, des images, des vidéos et de l’audio avec seulement ≈1 % de paramètres ajustés
arxiv.org
.
Alignement centré LLM : Les auteurs proposent un mécanisme de regroupement qui transforme des features patch‑level (images, vidéos, audio) en tokens de concepts et les alignent avec le langage via des tâches X‑to‑text sur Webvid‑2M, CC3M et AudioCaps
arxiv.org
. Cela améliore la cohérence sémantique entre les entrées multimodales et le LLM.
Instruction tuning multimodale (MosIT) : Un jeu de 5 000 dialogues multitors alternant les modalités est généré à l’aide de GPT‑4 et vérifié manuellement pour permettre un apprentissage d’instructions cross‑modal
arxiv.org
. Ce corpus favorise la compréhension des intentions implicites et la génération de contenus divers.
Amélioration des tâches de perception : NExT‑GPT obtient des scores CIDEr supérieurs ou comparables aux baselines sur NoCaps, Flickr30K et COCO (jusqu’à 124,9) et dépasse InstructBLIP, LLaVA et mPLUG‑Owl sur les questions visuelles VQAv2, VizWiz, OKVQA, MMB et SEED
arxiv.org
.
Performances vidéo/audio : Sur MSR‑VTT et MSVD‑QA, NExT‑GPT atteint 76,2 et 64,5 % respectivement, surpassant Video‑LLaVA et UIO‑2XXL ; sur MSRVTT‑QA et NExTQA, il atteint 61,4 et 50,7 %
arxiv.org
. Sur AudioCaps, il obtient 81,3 % (fine‑tune), montrant une capacité audio robuste.
Qualité de génération multimodale : Les scores FID, FAD et CLIPSIM démontrent que NExT‑GPT produit des images, des audios et des vidéos de meilleure qualité et plus cohérents que les modèles de diffusion autonomes (SD‑1.5, AudioLDM) et que CoDi ou UIO‑2XXL
arxiv.org
.
Ablation sur les tokens signal : Le nombre de tokens signal nécessaires varie selon la modalité : vidéo (12 tokens) exige plus d’informations tandis que l’image et l’audio atteignent des résultats satisfaisants avec 4 et 8 tokens
arxiv.org
. Cela guide le réglage de la granularité des signaux pour chaque modalité.
Importance du mécanisme de regroupement : Un remplacement du regroupement par un simple réseau linéaire dégrade les performances (VQAv2 66,7→63,8 ; VizWiz 48,4→45,4) et une Q‑former n’apporte qu’une légère amélioration, confirmant l’efficacité du regroupement pour l’alignement
arxiv.org
