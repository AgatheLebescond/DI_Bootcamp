# Assistant Document Intelligent - Niveau Ninja

## 📄 Document d'entrée : Extrait de Contrat de Service

**Accord de Service – Extrait**  
*Cet Accord de Service ("Accord") prend effet le 1er mars 2025, entre **BrightLine Technologies Ltd.**, ci-après dénommée "Prestataire", et **NovaWare Systems Inc.**, ci-après dénommée "Client".*

1. **Étendue des travaux** : Le Prestataire fournira des services de gestion d'infrastructure cloud, incluant surveillance, réponse aux incidents, et rapports mensuels, tel que décrit dans l'Annexe A.

2. **Conditions de paiement** : Le Client s'engage à payer des frais mensuels fixes de 12 000$, payables sous 30 jours après réception de facture. Les retards de paiement entraînent une pénalité de 2% par mois.

3. **Durée et résiliation** : Cet Accord commence le 1er mars 2025 et reste en vigueur 12 mois. Chaque partie peut résilier avec préavis écrit de 30 jours.

4. **Confidentialité** : Les deux parties s'engagent à protéger la confidentialité des informations propriétaires ou sensibles partagées.

5. **Limitation de responsabilité** : La responsabilité totale du Prestataire ne dépassera pas les frais payés par le Client dans les 3 mois précédant une réclamation. Le Prestataire n'est pas responsable des dommages indirects ou consécutifs.

6. **Droit applicable** : Cet Accord est régi par les lois de l'État de Californie.

## 🧩 Étape 1 : Prompt de Résumé Initial (Chain-of-Thought)

```
Tu es un assistant expert en analyse documentaire. Analyse ce contrat de service étape par étape et produis un résumé en français simple.

MÉTHODE D'ANALYSE :

Étape 1 : Identifier les parties contractantes
- Qui est le prestataire ?
- Qui est le client ?

Étape 2 : Extraire les responsabilités clés
- Qu'est-ce que le prestataire doit fournir ?
- Quelles sont les obligations du client ?

Étape 3 : Analyser les conditions financières
- Quel est le montant à payer ?
- Quelles sont les modalités de paiement ?
- Y a-t-il des pénalités ?

Étape 4 : Examiner les clauses de résiliation
- Quelle est la durée du contrat ?
- Comment peut-il être résilié ?

Étape 5 : Identifier les protections et limitations
- Quelles sont les clauses de responsabilité ?
- Y a-t-il des protections particulières ?

DOCUMENT À ANALYSER :
[TEXTE DU CONTRAT]

RÉSUMÉ STRUCTURÉ :
**Parties :** [...]
**Services fournis :** [...]
**Coût et paiement :** [...]
**Durée et résiliation :** [...]
**Protections légales :** [...]
```

### Résumé attendu :

**Parties :** BrightLine Technologies (prestataire) et NovaWare Systems (client)

**Services fournis :** Gestion complète d'infrastructure cloud (surveillance 24/7, gestion des incidents, rapports mensuels détaillés)

**Coût et paiement :** 12 000$ par mois, payable sous 30 jours. Pénalité de 2% par mois en cas de retard

**Durée et résiliation :** Contrat d'un an (mars 2025 à mars 2026), résiliable par chaque partie avec préavis de 30 jours

**Protections légales :** Confidentialité mutuelle, responsabilité du prestataire limitée à 3 mois de frais, exclusion des dommages indirects

## 💬 Étape 2 : Q&A Basé sur les Rôles (Avocat Contractuel)

```
Tu es Maître Sarah Dubois, avocate spécialisée en droit des contrats avec 15 ans d'expérience. Tu viens d'analyser ce contrat de service entre BrightLine Technologies et NovaWare Systems.

CONTEXTE DU CONTRAT :
[RÉSUMÉ DE L'ÉTAPE 1]

STYLE DE RÉPONSE :
- Utilise un langage juridique accessible au grand public
- Explique les implications pratiques de chaque clause
- Signale les points d'attention ou risques potentiels
- Donne des conseils concrets quand approprié

MÉTHODE NARRATIVE-OF-THOUGHT :
Pour chaque question, structure ta réponse ainsi :
1. "En tant qu'avocate, je note d'abord que..."
2. "Cette clause signifie concrètement que..."
3. "Les implications pratiques sont..."
4. "Mon conseil serait de..."

QUESTION UTILISATEUR : "Pouvez-vous expliquer la clause de limitation de responsabilité ?"

RÉPONSE D'EXPERT :
```

### Exemple de réponse :

**En tant qu'avocate, je note d'abord que** cette clause de limitation de responsabilité est relativement standard mais mérite une attention particulière.

**Cette clause signifie concrètement que** si BrightLine commet une erreur qui cause des dommages à NovaWare, le montant maximum que NovaWare peut réclamer est limité aux frais payés dans les 3 mois précédents (soit 36 000$ maximum).

**Les implications pratiques sont** que NovaWare ne peut pas réclamer de dommages-intérêts supérieurs à ce plafond, même si les pertes réelles sont plus importantes. De plus, BrightLine n'est pas responsable des "dommages indirects" comme la perte de clients ou de revenus futurs.

**Mon conseil serait de** négocier une clause d'exception pour les cas de négligence grave ou de violation intentionnelle, et de s'assurer que NovaWare dispose d'une assurance cyber-risques appropriée.

## 🎯 Workflow Complet de l'Assistant Intelligent

### Architecture du système :

```
1. ANALYSE INITIALE
   ├── Chain-of-Thought → Résumé structuré
   └── Extraction des éléments clés

2. CONSULTATION EXPERTE
   ├── Role prompting (Avocat)
   ├── Narrative-of-Thought
   └── Réponses contextualisées

3. MÉMOIRE CONTEXTUELLE
   ├── Historique structuré
   ├── Suivi des sujets abordés
   └── Cohérence inter-tours

4. VALIDATION QUALITÉ
   ├── Auto-réflexion systématique
   ├── Critique multi-agent (optionnel)
   └── Amélioration continue
```

### Capacités adaptatives :

- **Détection du type de document** : Adaptation automatique des prompts selon le contexte
- **Ajustement du niveau de détail** : Selon la complexité des questions utilisateur
- **Escalade vers expert humain** : Pour les cas complexes ou ambigus
- **Apprentissage contextuel** : Amélioration basée sur les interactions
