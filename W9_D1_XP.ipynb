# Exercices de Prompt Engineering - Solutions

## 🌟 Exercice 1 : Réécrire et Optimiser un Prompt Vague

### Prompt original problématique :
*"Écris quelque chose sur les conseils de productivité."*

### 1. Trois problèmes critiques identifiés :

1. **Manque de contexte de marque** - Aucune mention de FlowNest ou de positionnement spécifique
2. **Audience non définie** - Qui sont les utilisateurs cibles ? Entrepreneurs ? Étudiants ? Télétravailleurs ?
3. **Format inadapté à LinkedIn** - Pas de structure engageante, de call-to-action, ou de longueur optimale

### 2. Prompt réécrit avec les trois techniques :

```
Agis comme un expert en marketing digital spécialisé dans les applications de productivité.

Rédige un post LinkedIn engageant pour promouvoir FlowNest, notre nouvelle application de focus et concentration. 

INSTRUCTIONS :
- Donne 3 conseils pratiques de productivité que FlowNest aide à appliquer
- Utilise un ton professionnel mais accessible, avec des émojis pertinents
- Structure en : accroche + 3 conseils numérotés + call-to-action vers FlowNest
- Maximum 200 mots pour optimiser l'engagement LinkedIn

CONTRAINTES :
- Cible : professionnels et entrepreneurs de 25-45 ans
- Inclus 2-3 hashtags pertinents (#productivité #focus #entrepreneur)
- Termine par une question pour encourager les commentaires

Utilise ce format à chaque fois que nous promouvons des outils de productivité sur les réseaux sociaux.
```

## 🌟 Exercice 2 : Prompt Multi-Parties pour Génération de Quiz

### Prompt complet pour le contenu éducatif :

```
Tu es un enseignant de sciences expérimenté spécialisé dans la création de contenu pédagogique pour des élèves de 5ème (11-13 ans).

À partir de l'article sur les éruptions volcaniques fourni, crée du contenu éducatif suivant ce format :

**RÉSUMÉ (pour Google Slides) :**
• [Point clé 1 en langage simple]
• [Point clé 2 en langage simple]

**QUIZ INTERACTIF :**
Génère 3 questions à choix multiples, chacune avec :
- 1 bonne réponse
- 2 distracteurs plausibles mais incorrects
- Vocabulaire adapté aux 11-13 ans

**CONTRAINTES :**
- Ton amical et encourageant
- Évite le jargon scientifique complexe
- Questions basées uniquement sur l'article fourni
- Format prêt à copier-coller dans Google Slides

Exemple de format souhaité :
"Question 1 : Que se passe-t-il quand le magma remonte à la surface ?
A) Il devient de la lave
B) Il se transforme en roche
C) Il disparaît"
```

### Prompt vague comparé :
*"Fais un quiz pour enfants sur cet article."*

### Trois améliorations clés qui évitent les sorties de faible qualité :

1. **Spécification de l'âge et du niveau** - Évite un contenu trop complexe ou trop simple
2. **Structure imposée** - Garantit un format utilisable et cohérent
3. **Contraintes de vocabulaire** - Assure l'adaptation au public cible

## 🌟 Exercice 3 : Ajouter du Contexte

### 1. Analyse du contexte manquant :

| Type de Contexte | Manquant ? | Ce qui devrait être ajouté |
|------------------|------------|----------------------------|
| Rôle | Oui | Analyste financier ou assistant exécutif |
| Audience | Oui | Équipe dirigeante non-technique |
| Objectif | Oui | Présentation de 3 minutes, points d'action |
| Source | Oui | Type de rapport (mensuel, trimestriel, etc.) |
| Format/Style | Oui | Structure en bullet points, chiffres clés |
| Contraintes | Oui | Durée, niveau de détail, métriques prioritaires |

### 2. Prompt réécrit :

```
Agis comme un analyste financier senior préparant un briefing exécutif.

Cette synthèse est destinée à une équipe dirigeante non-technique pour une présentation mensuelle de 3 minutes.

TÂCHE : Résume ce rapport financier mensuel en te concentrant sur :

FORMAT REQUIS :
• 3 points clés maximum, chacun avec données chiffrées de support
• Structure : Situation actuelle → Impact → Action recommandée
• Langage accessible, évite le jargon technique

CONTRAINTES :
• Maximum 150 mots total
• Inclus les métriques les plus critiques (revenus, coûts, marge)
• Signale tout écart significatif par rapport aux objectifs
• Propose 1 action prioritaire si nécessaire

Le format doit permettre une présentation fluide de 3 minutes maximum.
```

## 🌟 Exercice 4 : Style Structuré pour Chatbot Support Client

### Style choisi : **Fonctionnel**

```
Tu es un assistant support client expert pour notre boutique e-commerce.

PROTOCOLE DE RÉSOLUTION :
1. Identifie le type de demande (commande, livraison, retour, produit)
2. Collecte les informations nécessaires (numéro de commande, email, etc.)
3. Fournis une solution précise avec étapes claires
4. Propose des alternatives si la solution principale n'est pas applicable
5. Confirme la satisfaction et oriente vers ressources additionnelles si besoin

CONTRAINTES :
• Ton professionnel mais chaleureux
• Réponses factuelles basées sur nos politiques uniquement
• Si tu ne sais pas : "Je vais vous mettre en relation avec un spécialiste"
• Maximum 100 mots par réponse initiale

ESCALADE : Transfère vers humain si : réclamation complexe, demande de remboursement >100€, client insatisfait après 2 échanges.
```

### Deux caractéristiques clés du style fonctionnel :
1. **Processus étape par étape** - Assure une résolution systématique
2. **Critères d'escalade précis** - Évite les blocages et améliore l'efficacité

### Justification du choix :
Le style fonctionnel est optimal ici car le support client nécessite des réponses précises, cohérentes et orientées solution. Contrairement au style conversationnel qui pourrait dériver, ou à l'exploratoire qui générerait trop d'options, le style fonctionnel garantit une expérience client standardisée et efficace.

## 🌟 Exercice 5 : Contrôle du Style, Structure et Longueur

### 1. Prompt avec contraintes strictes :

```
Rédige EXACTEMENT 3 bullet points pour promouvoir la smartwatch PulseOne Mini.

CONTRAINTES STRICTES :
• MAXIMUM 50 mots au total (compte chaque mot)
• Ton amical et enthousiaste
• DOIT mentionner : autonomie batterie, suivi fitness, compatibilité Bluetooth
• Format : • [Point 1] • [Point 2] • [Point 3]

VÉRIFICATION OBLIGATOIRE :
- Compte tes mots avant de répondre
- Vérifie que les 3 caractéristiques sont présentes
- Assure-toi du ton amical

NE DÉPASSE JAMAIS 50 mots. Si c'est trop long, raccourcis sans perdre les éléments obligatoires.
```

### 2. Critères d'évaluation :
- ✅ Respect des 50 mots maximum
- ✅ Mention des 3 caractéristiques (batterie, fitness, Bluetooth)
- ✅ Ton amical et engageant
- ✅ Format en bullet points

### 3. Révision si nécessaire :
*"IMPÉRATIF : Ne dépasse JAMAIS 50 mots. Compte précisément. Liste EXACTEMENT 3 bullet points. Inclus obligatoirement : autonomie, fitness, Bluetooth."*

## 🌟 Exercice 6 : Détection et Atténuation des Hallucinations

### 1. Prompt anti-hallucination :

```
Tu es un assistant de recherche scientifique rigoureux.

TÂCHE : Résume cet article sur le changement climatique et la biodiversité marine.

RÈGLES STRICTES ANTI-HALLUCINATION :
• N'inclus QUE les informations explicitement mentionnées dans l'article fourni
• N'ajoute aucune donnée externe, statistique ou fait non présent dans le texte
• Si une information te semble incomplète, écris : "L'article ne précise pas..."
• N'interpole pas de données entre différentes études

VÉRIFICATION OBLIGATOIRE :
Avant chaque affirmation, demande-toi : "Cette information est-elle mot pour mot dans l'article ?"

Si tu n'es pas certain, utilise : "Selon l'article..." ou "L'étude mentionne que..."

INTERDIT : Statistiques, pourcentages, ou prédictions non explicitement cités dans le document source.
```

### 2. Prompt avec langage de vérification :

```
SYNTHÈSE VÉRIFIÉE - Article biodiversité marine

INSTRUCTIONS DE VÉRIFICATION :
• Inclus uniquement les affirmations explicitement énoncées dans l'article
• Pour chaque fait : cite la section/paragraphe source
• Si l'information n'est pas disponible dans le texte, indique : "Information non fournie dans l'article"
• Distingue clairement les faits des hypothèses de l'auteur

FORMAT DE CITATION :
"Selon l'article, [fait exact]" ou "L'étude indique que [citation directe]"

En cas d'incertitude sur une information, OMETS-LA plutôt que de risquer une erreur.
```

### 3. Deux stratégies d'atténuation utilisées :
1. **Vérification systématique** - Obligation de vérifier chaque affirmation contre le texte source
2. **Langage conditionnel** - Utilisation de formules comme "selon l'article" pour ancrer dans la source

### 4. Domaines à risque élevé :

**Domaine juridique :**
- Risque : Mauvaise interprétation de jurisprudence ou de lois
- Impact : Conseils juridiques erronés pouvant avoir des conséquences légales graves

**Domaine médical :**
- Risque : Diagnostic incorrect ou traitement inapproprié suggéré
- Impact : Mise en danger de la santé du patient, retard de traitement approprié

## 🎯 Points clés à retenir

### Techniques essentielles de prompt engineering :
1. **Définir le rôle** ("Agis comme un...")
2. **Préciser l'audience** (niveau, contexte, besoins)
3. **Structurer les instructions** (format, contraintes, étapes)
4. **Prévenir les hallucinations** (sources uniquement, vérification)
5. **Optimiser pour l'usage** (réutilisabilité, adaptation)

### Bonnes pratiques :
- Toujours tester et itérer les prompts
- Adapter le style au contexte d'usage
- Inclure des exemples quand nécessaire
- Prévoir les cas d'échec et les alternatives
