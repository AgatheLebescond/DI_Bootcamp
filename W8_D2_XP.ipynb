# Exercice 1 : Réflexion sur les niveaux Open Source

## Qu'allez-vous utiliser

- **Niveaux open source** : entièrement ouvert, poids publiés, architecture uniquement  
- **Composants du modèle** : architecture du modèle, poids pré-entraînés  
- **Concepts de recyclage** : réglage fin, adaptation au domaine  
- **Exigences du domaine** : sensibilité des données de santé, conformité  

---

## Instructions étape par étape

### 1. Rassembler les définitions

Localisez les descriptions des trois niveaux d’ouverture :  
- **Entièrement ouvert**  
- **Poids libérés**  
- **Architecture uniquement**  

Copiez ou surlignez chaque définition dans vos notes.

---

### 2. Identifier les caractéristiques clés

Pour chaque niveau, notez deux points :  
- **Qu'est-ce qui est ouvert ?** (ex. : code, pondérations, les deux)  
- **Ce que vous pouvez et ne pouvez pas faire** (ex. : vous recycler, inspecter les éléments internes)

---

### 3. Comparez-les côte à côte

| **Niveau d’ouverture** | **Qu'est-ce qui est ouvert ?** | **Impact sur la reconversion/modification** |
|------------------------|-------------------------------|----------------------------------------------|
| Entièrement ouvert     | Code + poids + architecture   | Vous pouvez inspecter, modifier et recycler l’ensemble du modèle. |
| Poids libérés          | Poids uniquement              | Vous disposez des poids pour un réglage fin, mais pas du code complet. |
| Architecture uniquement| Structure uniquement          | Vous connaissez la structure mais vous manquez de poids pré-entraînés. |

---

### 4. Rédigez votre paragraphe comparatif

Rédigez un paragraphe de 3 à 5 phrases qui :  
- Définit chaque niveau d’ouverture avec vos propres mots.  
- Souligne un avantage ou une limitation principale de chacun.  

---

### 5. Répondez à l'invite sur les soins de santé

Réfléchissez à la création d’un assistant spécifique aux soins de santé qui doit être recyclé sur les données cliniques.  
Écrivez 1 à 2 phrases expliquant :  
- **Quel niveau d’ouverture est essentiel pour cette tâche.**  
- **Pourquoi ce niveau prend en charge le réglage fin spécifique au domaine.**

---

## Livrables

- **Paragraphe comparatif** (3 à 5 phrases)  
- **Réponse rapide** (1 à 2 phrases)


Voici la réponse à l’exercice 2, de façon directe et complète :

**Paragraphe comparatif**
Un modèle **entièrement ouvert** met à disposition à la fois l’architecture et les poids pré-entraînés, ainsi que le code source. Il permet une inspection complète, une modification du modèle et un réglage fin. Un modèle avec **poids libérés** fournit les poids pré-entraînés mais pas nécessairement le code complet, permettant un réglage fin mais limitant la compréhension ou la modification de l’architecture. En revanche, un modèle avec **architecture uniquement** expose seulement la structure du réseau, sans poids pré-entraînés, ce qui empêche tout ajustement direct et nécessite un nouvel entraînement complet. Chaque niveau offre un degré différent de contrôle et d’adaptabilité.

**Réponse rapide sur les soins de santé**
Le niveau **entièrement ouvert** est essentiel pour créer un assistant médical, car il permet un réglage fin sur des données cliniques sensibles tout en assurant transparence et conformité.

# Applications potentielles innovatives et créatives

## 1. Assistant médical localisé et conforme (Healthcare Assistant)

**Objectif :**
Un assistant conversationnel spécialisé en santé respectant la sensibilité des données cliniques et déployable localement. ([arXiv][1])

**Méthode & outils :**

* Utiliser un modèle **entièrement ouvert** (code + poids) comme Mistral 7B Instruct sous licence Apache 2.0 ([ProjectPro][2]).

* Déploiement local via `llama.cpp` en quantification 4 bits (`GGUF`) pour fonctionner sur CPU/WSL2 ([arXiv][1]).

**Faisabilité :**
La configuration locale (≥ 16 Go RAM, déployé via Docker ou WSL2) couvre les prérequis ([ai-academy.com][3]). Coût nul de cloud et contrôle total des données.
