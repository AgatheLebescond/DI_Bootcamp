```python
#!/usr/bin/env python3
"""
AI Greeter - Application de Salutation Personnalisée
Daily Challenge: Créer votre premier AI Greeter
"""

import subprocess
import sys
from datetime import datetime

# Installation automatique des dépendances
def install_dependencies():
    """Installe les dépendances nécessaires"""
    packages = ["gradio", "streamlit"]
    
    for package in packages:
        try:
            __import__(package)
        except ImportError:
            print(f"📦 Installation de {package}...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", package])
    
    print("✅ Dépendances installées avec succès")

# Installation
install_dependencies()

import gradio as gr
import streamlit as st

# ================================================================
# ✅ ÉTAPE 1: FONCTION PYTHON POUR SALUTATION PERSONNALISÉE
# ================================================================

def create_personalized_greeting(name, language):
    """
    Crée une salutation personnalisée basée sur le nom, la langue et l'heure
    
    Args:
        name (str): Nom de l'utilisateur
        language (str): Langue choisie ("English" ou "French")
    
    Returns:
        str: Message de salutation personnalisé
    """
    # Obtenir l'heure actuelle
    current_time = datetime.now()
    hour = current_time.hour
    
    # Détermination du moment de la journée
    if 5 <= hour < 12:
        time_period = "morning"
    elif 12 <= hour < 17:
        time_period = "afternoon" 
    elif 17 <= hour < 22:
        time_period = "evening"
    else:
        time_period = "night"
    
    # Messages selon la langue et l'heure
    greetings = {
        "English": {
            "morning": f"Good morning, {name}! ☀️",
            "afternoon": f"Good afternoon, {name}! 🌞", 
            "evening": f"Good evening, {name}! 🌅",
            "night": f"Good night, {name}! 🌙"
        },
        "French": {
            "morning": f"Bonjour, {name} ! ☀️",
            "afternoon": f"Bon après-midi, {name} ! 🌞",
            "evening": f"Bonsoir, {name} ! 🌅", 
            "night": f"Bonne nuit, {name} ! 🌙"
        }
    }
    
    # Validation des entrées
    if not name.strip():
        if language == "French":
            return "❌ Veuillez entrer votre nom pour recevoir une salutation personnalisée !"
        else:
            return "❌ Please enter your name to receive a personalized greeting!"
    
    # Génération du message de salutation
    greeting = greetings.get(language, greetings["English"]).get(time_period, "Hello!")
    
    # Ajout d'informations contextuelles
    time_info = current_time.strftime("%H:%M")
    
    if language == "French":
        context = f"\n\n🕐 Il est actuellement {time_info}. J'espère que vous passez une excellente journée !"
    else:
        context = f"\n\n🕐 It's currently {time_info}. Hope you're having a wonderful day!"
    
    return greeting + context

# ================================================================
# ✅ ÉTAPE 2: INTERFACE GRADIO
# ================================================================

def create_gradio_interface():
    """Crée l'interface Gradio pour l'AI Greeter"""
    
    # Création de l'interface avec gr.Interface()
    interface = gr.Interface(
        fn=create_personalized_greeting,
        inputs=[
            gr.Textbox(
                label="👤 Votre Nom",
                placeholder="Entrez votre nom ici...",
                info="Tapez votre prénom ou nom complet"
            ),
            gr.Dropdown(
                choices=["English", "French"],
                label="🌐 Langue de Salutation",
                value="French",
                info="Choisissez la langue pour votre salutation"
            )
        ],
        outputs=gr.Textbox(
            label="🎉 Votre Salutation Personnalisée",
            info="Votre message de bienvenue apparaîtra ici"
        ),
        title="🤖 AI Greeter - Salutations Personnalisées",
        description="""
        **Bienvenue dans l'AI Greeter !** 🎊
        
        Cette application vous accueille chaleureusement en tenant compte de :
        • Votre nom personnel
        • L'heure actuelle de la journée
        • Votre langue préférée
        
        Parfait pour une plateforme communautaire ou un site web accueillant !
        """,
        
        # ✅ ÉTAPE 3: EXEMPLES POUR TESTS RAPIDES
        examples=[
            ["Alice", "English"],
            ["Jean", "French"],
            ["Marie", "French"],
            ["John", "English"],
            ["Pierre", "French"]
        ],
        
        theme=gr.themes.Soft(),
        allow_flagging="never",
        show_share_button=True
    )
    
    return interface

# ================================================================
# VERSION STREAMLIT ALTERNATIVE
# ================================================================

def create_streamlit_app():
    """Code pour une version Streamlit de l'AI Greeter"""
    
    streamlit_code = '''
import streamlit as st
from datetime import datetime

# Configuration de la page
st.set_page_config(
    page_title="AI Greeter",
    page_icon="🤖",
    layout="centered"
)

# Fonction de salutation (même que Gradio)
def create_personalized_greeting(name, language):
    current_time = datetime.now()
    hour = current_time.hour
    
    if 5 <= hour < 12:
        time_period = "morning"
    elif 12 <= hour < 17:
        time_period = "afternoon" 
    elif 17 <= hour < 22:
        time_period = "evening"
    else:
        time_period = "night"
    
    greetings = {
        "English": {
            "morning": f"Good morning, {name}! ☀️",
            "afternoon": f"Good afternoon, {name}! 🌞", 
            "evening": f"Good evening, {name}! 🌅",
            "night": f"Good night, {name}! 🌙"
        },
        "French": {
            "morning": f"Bonjour, {name} ! ☀️",
            "afternoon": f"Bon après-midi, {name} ! 🌞",
            "evening": f"Bonsoir, {name} ! 🌅", 
            "night": f"Bonne nuit, {name} ! 🌙"
        }
    }
    
    if not name.strip():
        if language == "French":
            return "❌ Veuillez entrer votre nom !"
        else:
            return "❌ Please enter your name!"
    
    greeting = greetings.get(language, greetings["English"]).get(time_period, "Hello!")
    time_info = current_time.strftime("%H:%M")
    
    if language == "French":
        context = f"\\n\\n🕐 Il est {time_info}. Excellente journée !"
    else:
        context = f"\\n\\n🕐 It's {time_info}. Have a great day!"
    
    return greeting + context

# Interface Streamlit
st.title("🤖 AI Greeter - Salutations Personnalisées")
st.markdown("**Bienvenue !** Cette app vous salue selon votre nom, l'heure et votre langue.")

# Inputs utilisateur
col1, col2 = st.columns(2)

with col1:
    user_name = st.text_input(
        "👤 Votre Nom",
        placeholder="Entrez votre nom...",
        help="Tapez votre prénom ou nom complet"
    )

with col2:
    language = st.selectbox(
        "🌐 Langue",
        ["French", "English"],
        help="Choisissez la langue pour votre salutation"
    )

# Bouton et affichage
if st.button("🎉 Générer ma Salutation", type="primary"):
    if user_name:
        greeting = create_personalized_greeting(user_name, language)
        st.success(greeting)
    else:
        st.error("Veuillez entrer votre nom d'abord !")

# Exemples en sidebar
st.sidebar.header("💡 Exemples Rapides")
examples = [
    {"name": "Alice", "lang": "English"},
    {"name": "Jean", "lang": "French"},
    {"name": "Marie", "lang": "French"},
    {"name": "John", "lang": "English"}
]

for ex in examples:
    if st.sidebar.button(f"{ex['name']} ({ex['lang']})"):
        greeting = create_personalized_greeting(ex['name'], ex['lang'])
        st.success(greeting)

# Chat version avec st.chat_message
st.header("💬 Version Chat")

if "messages" not in st.session_state:
    st.session_state.messages = []

for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.write(message["content"])

if prompt := st.chat_input("Tapez votre nom pour une salutation..."):
    # Message utilisateur
    with st.chat_message("user"):
        st.write(f"Je m'appelle {prompt}")
    
    st.session_state.messages.append({"role": "user", "content": f"Je m'appelle {prompt}"})
    
    # Réponse AI Greeter
    greeting = create_personalized_greeting(prompt, language)
    
    with st.chat_message("assistant"):
        st.write(greeting)
    
    st.session_state.messages.append({"role": "assistant", "content": greeting})
'''
    
    return streamlit_code

# ================================================================
# FONCTION PRINCIPALE ET MENU
# ================================================================

def main():
    """Fonction principale avec menu de choix"""
    
    print("🤖 AI Greeter - Application de Salutation Personnalisée")
    print("=" * 60)
    print("Daily Challenge: Build Your First AI Greeter")
    
    while True:
        print("\n📋 Choisissez votre interface :")
        print("1. 🎨 Gradio Interface (Recommandé)")
        print("2. 📱 Streamlit Chat Version")
        print("3. 🧪 Test de la fonction Python")
        print("4. 📝 Générer fichier Streamlit")
        print("0. ❌ Quitter")
        
        choice = input("\n👉 Votre choix (0-4) : ").strip()
        
        if choice == "1":
            print("\n🎨 Lancement de l'interface Gradio...")
            print("🌐 L'interface va s'ouvrir dans votre navigateur")
            
            # ✅ ÉTAPE 4: LANCEMENT LOCAL POUR TEST
            interface = create_gradio_interface()
            interface.launch(
                share=False,
                inbrowser=True,
                show_error=True,
                quiet=False
            )
        
        elif choice == "2":
            print("\n📱 Code Streamlit généré !")
            print("Copiez ce code dans 'ai_greeter_streamlit.py' :")
            print("-" * 50)
            print(create_streamlit_app())
            print("-" * 50)
            print("Puis lancez : streamlit run ai_greeter_streamlit.py")
        
        elif choice == "3":
            print("\n🧪 Test de la fonction Python...")
            test_function()
        
        elif choice == "4":
            generate_streamlit_file()
        
        elif choice == "0":
            print("👋 Au revoir ! Merci d'avoir testé l'AI Greeter !")
            break
        
        else:
            print("❌ Choix invalide, veuillez réessayer.")

def test_function():
    """Teste la fonction de salutation Python"""
    
    print("\n🧪 Tests de la fonction create_personalized_greeting():")
    print("-" * 50)
    
    # Tests avec les exemples de l'énoncé
    test_cases = [
        ("Alice", "English"),
        ("Jean", "French"),
        ("Marie", "French"),
        ("John", "English"),
        ("", "French"),  # Test nom vide
    ]
    
    for name, language in test_cases:
        result = create_personalized_greeting(name, language)
        print(f"✅ Nom: '{name}', Langue: {language}")
        print(f"   Résultat: {result}")
        print()

def generate_streamlit_file():
    """Génère le fichier Streamlit séparé"""
    
    filename = "ai_greeter_streamlit.py"
    
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            f.write(create_streamlit_app())
        
        print(f"✅ Fichier {filename} créé avec succès !")
        print(f"📱 Lancez avec : streamlit run {filename}")
        
    except Exception as e:
        print(f"❌ Erreur lors de la création du fichier: {e}")

# Démonstration interactive
def interactive_demo():
    """Démonstration interactive en ligne de commande"""
    
    print("\n🎯 Démonstration Interactive de l'AI Greeter")
    print("=" * 50)
    
    while True:
        name = input("👤 Entrez votre nom (ou 'quit' pour arrêter) : ").strip()
        
        if name.lower() == 'quit':
            break
        
        print("\n🌐 Langues disponibles :")
        print("1. Français")
        print("2. English")
        
        lang_choice = input("Choisissez (1 ou 2) : ").strip()
        language = "French" if lang_choice == "1" else "English"
        
        greeting = create_personalized_greeting(name, language)
        
        print(f"\n🎉 {greeting}")
        print("-" * 50)

if __name__ == "__main__":
    # Option pour demo interactive ou menu principal
    import sys
    
    if len(sys.argv) > 1 and sys.argv[1] == "--demo":
        interactive_demo()
    else:
        main()

# ================================================================
# INSTRUCTIONS ET DOCUMENTATION
# ================================================================
"""
🎯 AI GREETER - INSTRUCTIONS COMPLÈTES

📦 Installation :
pip install gradio streamlit

🚀 Lancement :
python ai_greeter.py

✅ Fonctionnalités implémentées :

ÉTAPE 1 - Fonction Python ✅
• create_personalized_greeting(name, language)
• Utilise datetime pour détecter l'heure
• Retourne salutations selon langue et moment :
  - "Good morning, Alice!" ou "Bonjour, Alice!"
  - Adapté pour matin/après-midi/soir/nuit

ÉTAPE 2 - Interface Gradio ✅  
• gr.Interface() avec inputs texte + dropdown
• Champs : nom (texte), langue (dropdown)
• Output : greeting personnalisé
• Design professionnel avec thème Soft

ÉTAPE 3 - Exemples pour tests ✅
• examples=[["Alice", "English"], ["Jean", "French"]]
• + exemples supplémentaires pour démo complète

ÉTAPE 4 - Test local navigateur ✅
• Interface.launch() avec inbrowser=True
• Accessible sur localhost avec interface web
• Partage optionnel disponible

🎨 Fonctionnalités bonus :
• Version Streamlit avec st.chat_message
• Mode démonstration interactive
• Tests automatisés de la fonction
• Messages contextuels avec heure
• Validation des entrées utilisateur
• Emojis selon moment de la journée

🧪 Tests inclus :
• Noms français et anglais
• Gestion des entrées vides
• Tous moments de la journée
• Validation des langues

Parfait pour plateforme communautaire ! 🎊
"""
```
