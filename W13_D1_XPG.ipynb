````python
#############################################
# FICHIER: server_stream.py
#############################################

# Ce script lance un serveur MCP en HTTP streamable (localhost), 
# expose une fonction longue qui émet des notifications structurées de progression 
# (sans jamais mélanger notifications et résultat final).

import asyncio
try:
    from mcp.server.fastmcp import FastMCP
except ImportError:
    # Selon ta version SDK, adapte l'import, ex: from mcp.server import FastMCP
    raise

mcp = FastMCP("Streaming-Demo")

@mcp.tool(description="Simule un traitement long avec notifications de progression")
async def process_items(total: int = 5, ctx=None) -> str:
    """
    Simule un traitement long et émet une notification structurée à chaque étape.
    Le message final est renvoyé UNE SEULE FOIS à la fin (MCP contract).
    """
    for i in range(1, total + 1):
        # On simule un travail avec un petit délai
        await asyncio.sleep(0.3)
        # Envoi d'une notification structurée, toujours via le contexte MCP
        if ctx is not None:
            # La méthode info() est standard MCP pour les notifications.
            ctx.info({"progress": f"item {i}/{total}"})
    # Retour unique, ne pas émettre d'autres notifications après.
    return f"Traitement terminé pour {total} items."

if __name__ == "__main__":
    # Privilégie Streamable HTTP (localhost SEULEMENT pour le dev)
    # ATTENTION: Pour exposer au-delà de localhost, il faut impérativement ajouter
    #           - une authentification (token dans headers)
    #           - CORS restrictif (allow-list des domaines)
    # Ajoute le code ici si besoin, selon ton SDK:
    try:
        # Selon SDK, le flag peut différer ("streamable-http" ou "http-stream")
        mcp.run(transport="streamable-http", host="127.0.0.1", port=8080)
    except Exception as e:
        print("HTTP streamable non supporté. Bascule en STDIO. (Documente ce choix si tu rends l'exercice)")
        mcp.run()  # STDIO fallback
````

````python
#############################################
# FICHIER: client_stream.py
#############################################

# Ce script connecte le client à ton serveur MCP (HTTP privilégié, STDIO fallback),
# intercepte les notifications de progression et affiche séparément le résultat final.

import asyncio
import sys

try:
    from mcp.client.http import http_client, HttpServerParameters
except ImportError:
    http_client = None

from mcp import ClientSession, types

try:
    from mcp.client.stdio import stdio_client, StdioServerParameters
except ImportError:
    # Selon version MCP, adapte l'import
    raise

# Handler de notifications et messages du serveur
async def on_message(message):
    """
    Handler qui traite toutes les notifications 'info' MCP et les affiche distinctement.
    On ne mélange jamais notification et résultat final.
    """
    # MCP recommande le champ 'type' ou 'kind'
    if hasattr(message, "kind") and message.kind == types.MessageKind.INFO:
        # Notification structurée, affichage immédiat
        print("NOTIFICATION:", message.data)
    # On peut ajouter ici d'autres traitements pour d'autres types de messages MCP si besoin.

async def run():
    # Essaie d'abord la connexion HTTP, sinon fallback STDIO
    use_http = http_client is not None
    if use_http:
        params = HttpServerParameters(
            url="http://127.0.0.1:8080",  # Pour prod: ne jamais ouvrir au public sans sécurité !
        )
        async with http_client(params) as (read, write):
            async with ClientSession(read, write, message_handler=on_message) as session:
                await session.initialize()
                # Appelle le tool long avec un argument (ex: total=5)
                result = await session.invoke("process_items", {"total": 5})
                print("RESULTAT FINAL:", result)
    else:
        # Fallback local avec STDIO, le client lance le serveur lui-même
        params = StdioServerParameters(server="python", args=["server_stream.py"])
        async with stdio_client(params) as (read, write):
            async with ClientSession(read, write, message_handler=on_message) as session:
                await session.initialize()
                result = await session.invoke("process_items", {"total": 5})
                print("RESULTAT FINAL:", result)

if __name__ == "__main__":
    # Lancement asynchrone compatible Jupyter ET VSCode
    if sys.platform == "win32":
        asyncio.set_event_loop_policy(asyncio.WindowsSelectorEventLoopPolicy())
    asyncio.run(run())
````
